2. Определите распределение функции Pclass. Теперь Для мужчин и женщин отдельно. Сколько людей из второго класса было на борту?

Распределение функции Pclass:

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

data = pd.read_csv('titanic_train.csv', index_col='PassengerId')
print(data['Pclass'].value_counts().sort_index())
```

Для мужчин и женщин отдельно:

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

data = pd.read_csv('titanic_train.csv', index_col='PassengerId')
print(pd.crosstab(data['Pclass'], data['Sex']))
```

3. Каковы медиана и стандартное отклонениеFare?. Округлите до 2-х знаков после запятой.

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

data = pd.read_csv('titanic_train.csv', index_col='PassengerId')
print(round(data['Fare'].median(), 2))
print(round(data['Fare'].std(), 2))
```

4. Правда ли, что средний возраст выживших людей выше, чем у пассажиров, которые в конечном итоге умерли? 

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt    

data = pd.read_csv('titanic_train.csv', index_col='PassengerId')
print(round(data[data['Survived'] == 1]['Age'].mean(), 2))
print(round(data[data['Survived'] == 0]['Age'].mean(), 2))
```

5. Это правда, что пассажиры моложе 30 лет. выжили чаще, чем те, кому больше 60 лет. Каковы доли выживших людей среди молодых и пожилых людей?

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

data = pd.read_csv('titanic_train.csv', index_col='PassengerId')
print(data[(data['Age'] < 30) & (data['Survived'] == 1)].shape[0])
print(data[(data['Age'] > 60) & (data['Survived'] == 1)].shape[0])
print(round(data[data['Age'] < 30]['Survived'].mean(), 2))
print(round(data[data['Age'] > 60]['Survived'].mean(), 2))
```

6. Правда ли, что женщины выживали чаще мужчин? Каковы доли выживших людей среди мужчин и женщин?
